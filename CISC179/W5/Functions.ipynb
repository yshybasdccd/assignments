# Create a unit conversion program using functions
1a. The user selects kilometers per liter (kpl), and the response will be provided in miles per gallon (mpg). The units must be interchangeable, so the program will ask the user whether to convert from kpl to mpg or vice versa.

The program will prompt the user for input and deliver output with the appropriate units.

Additionally, the program will include input validation. For example, it will not accept letter inputs and will provide an error message to the user when invalid input is detected.

The function will also allow multiple arguments, enabling the user to convert multiple values at once.

Research and find out the conversion factor between the units.

````python
# Write your code here.

inputList = []

def userInputs():
  unitInput = input("Enter input unit: 1. Kilometer per liter. 2. Miles per galon.")
  userInput = input("What is your car's gas usage?")
  return unitInput, userInput

def inputValidation(unit, userInput):
  if((unit == "1" or unit == "2") and userInput.isdigit()):
    return True
  print("Invalid input")
  return False

def convertMetricToUS(x):
  return x*2.35

def convertUSToMetric(x):
  return x*0.425

def convertFunction(inputList):
  for item in inputList:
    if(item[0] == "1"):
      print(item[1], 'kilometers per liter = ' , convertMetricToUS(item[1]), ' mpg.')
    elif(item[0] == "2"):
      print(item[1], 'mpg = ', convertUSToMetric(item[1]), ' kilometers per liter.')


while(True):
  unitInput, userInput = userInputs()
  if(inputValidation(unitInput, userInput) == True):
    inputList.append([unitInput,int(userInput)])
    continueInput = input("Do you want to continue? Y/N")
    if(continueInput.capitalize() == "N"):
      break



if(unitInput == "1"):
  convertFunction(inputList)
elif(unitInput == "2"):
  convertFunction(inputList)
````

1c. What would be the result of changing a list or dictionary that was passed into a function as a parameter value? Which operations would be likely to create changes that would be visible outside the function? What steps might you take to minimize that risk?

Explain the above statements with the help of code.

````python
# Write your code here.
def function(my_list):
  print(id(my_list))
  my_list.append(100)
  print(id(my_list))

my_list = [10,20,30,40]
print('List before function: ', my_list)
function(my_list)
print('List after function: ', my_list)

# The result is that the list has been changed. We can make a temporary list
# with a copy of original in the function to prevent this.

def function2(my_list):
  print(id(my_list))
  temp_list = my_list.copy()
  temp_list.append(100)
  print(id(temp_list))

my_list = [10,20,30,40]
print('List before function: ', my_list)
function2(my_list)
print('List after function: ', my_list)
````
